Encapsulamiento en PPOO Publico vs Privado

El encapsulamiento en python es un pilar de la programacion orientada a objetos que restringe el acceso directo a los atributos y metrodos de una clase para evitar modificaciones accidentales. Se logra mediante convenciones de nombres: un guion bajo _ para miembros protegidos y doble guion bajo __ para privados. limitando su visibilidad externa.

Peincipales conceptos de Encapsulamiento de Python:

-Publico (por defecto): Kis atributos y métodos son accesibles desde fuera de la clase.
-Protegido (_variable): Se usa un guion bajo antes del nombre. Es una convención que indica que ...
-Privado
-Getters y Setters: Metodos publicos utilizados para obtener (get) o modificar (set) el valor de ...

```python
class CuentaBancaria:
    def __init__(self, titular, saldo):
        self.titular = titular        # Público
        self.__saldo = saldo          # Privado (doble guion bajo)

    # Getter para saldo
    def get_saldo(self): # o getSaldo(self):
        return self.__saldo

    # Setter para saldo con validación
    def depositar(self, cantidad):
        if cantidad > 0:
            self.__saldo += cantidad
            return True
        else:
            return False

cuenta = CuentaBancaria("Juan", 1000)
#print(cuenta.__saldo)  # Esto generaría un AttributeError
print(cuenta.get_saldo())
cuenta.depositar(500)  #este sería un setter
print(cuenta.get_saldo())
```
