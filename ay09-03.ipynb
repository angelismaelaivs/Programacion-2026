Programa que solicite al usuatrio el número de materias que lleva en el semetres, para  posteriormente solicitar el nombre de cada materia y que el usuario pueda ingresar su calificacion. El programa debe devolver el promedio

```python
n = int(input(print('¿Cuántas materias llevas?')))
lista = [] #diccionario que vamos a ir almacenando
materias = {}
i = 0
while i < n:
  print('Igresa el nombre de la materia:')
  lista.append(input('Nombre de la materia:'))    #append añade cosas a una lista
  i =+ 1
s = 0
for i in lista:
  materias[i] = float(input(f'Dame la calificación de {i}:'))
  s += materias[i]

  print('Tu promedio es de: ', s/n)
```
