Calculo de recibo de un restaurante
Pide el costo de una comida. Calcula el IVA (%16) y la propina el %10 cobre el costo sin IVA. Imprime propina y total

```python
prin("---Calculadora de cuenta ---")

c_c = float(input("Ingresa el costo de la comida: $ "))

iva = c_c * 0.16
propina = c_c * 0.1
total =  c_c + iva + propina

print("---Desglose de la cuenta---")
print("Costo original: $",{c_c})
print("IVA:    ",{iva.2f})
#el .2f es para limitar los decimales
print("Propina:   ",{propina.2f})
print("TOTAL A PAGAR:   ",{total.2f})

```

Ejercicio Ordenando sin listas

Lee tres enteros del ususario, imprimelos ordenados de menor a mayor usando solo las funciones min, max y aritmetica basica suma resta no se vale usar listas ni if/else

```python

a = print(int(input("Introduce un numero entero:")))
b = print(int(input("Introduce un numero entero:")))
c = print(int(input("Introduce un numero entero:")))

minimo = min(a,b,c)
maximo = max(a,b,c)
medio = a + b + c - minimo - maximo

print(f"Los numeros ordenados de menos a mayor son:, {minimo}, {medio}, {maximo}")

```
