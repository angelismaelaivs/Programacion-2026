# Definicion Clase
Una clase ees una definicion de un conjutno de elementos que sea va a crear apartir de ella, donde se guardan los atributos que tienen los objetos de la clase o los procesos que conllevan.

```python
#cuenta.py
class Cuenta:
  def __init__  (self,a,b):
    self.saldo = a
    self.tipo = b

  def depositar(self,cantidad):
    self.saldo = self.saldo + cantidad

  def retirar(self,cantidad):
    self.saldo = self.saldo - cantidad

#menu.py
#from cuenta import *
class Menu:
      pass
print("Probando el objeto")
cuenta1 = Cuenta (3000,"D")
print(cuenta1.saldo)
print(cuenta1.tipo)

cuenta1.depositar(5000)

print("probando el deposito")
print(cuenta1.saldo)

cuenta1.retirar(1200)
print("probaddo el retiro")
print(cuenta1.saldo)
```
