```python
class Alumno:
  def __init__(self, nombre, apellido, edad, noMaterias):
    self.nombre = nombre
    self.apellido = apellido
    if isinstance(edad, int):
      self.__edad = edad
    else:
      raise ValueError("La edad debe ser un número entero.")

    if isinstance(noMaterias, int):
      self.__noMaterias = noMaterias
    else:
      raise ValueError("El número de materias debe ser un número entero.")


  def getEdad(self):
    return self.__edad

  def setEdad(self, nuevaEdad):
    if isinstance(nuevaEdad, int):
      if nuevaEdad >= 0:
        self.__edad = nuevaEdad
      else:
        print("La edad no puede ser negativa.")
    else:
      print("La edad debe ser un número entero.")


  def getNoMaterias(self):
    return self.__noMaterias

  def setNoMaterias(self, nuevasMaterias):
    if isinstance(nuevasMaterias, int):
      if nuevasMaterias >= 0:
        self.__noMaterias = nuevasMaterias
      else:
        print("El numero de materias debe ser positivo.")
    else:
      print("El numero de materias debe ser un número entero.")


  def __str__(self):
    return f"Alumno(Nombre: {self.nombre}, Apellido: {self.apellido}, Edad: {self.__edad}, Número de Materias: {self.__noMaterias})"
```
