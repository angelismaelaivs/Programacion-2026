Crea un programa que sume todos los numeros ingresados por el usuario, ignorando las lienas que no contengan numerosvalidos. El programa debe mostrar la suma acrual despues de cada numero ingresado. Debe mostrar un mensaje de error apropiado tras cualquier entrada no valida y a continuacion, continuar sumando los numeros adicionales ingresados por el usuario. El programa debe finalizar cuando el usuario ingrese una linea en blanco. Asegurese de uqe el programa funcione correctamente tanto con numeros enteros como con numeros de coma flotante.

```python
valor = float(input('Ingresa un numero a sumar:'))
suma = 0

while valor != "":
  try:
    suma = suma + valor
    print(f'El total de la nueva suma es:{suma}')
    valor = float(input('Ingresa un numero a sumar:'))
  except ValueError:
    print('El valor que ingresaste no es un nùmero.')



print(f'La suma total es: {suma}')


```
